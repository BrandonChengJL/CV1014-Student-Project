```python
import pandas as pd
import re
import tkinter as tk
from tkinter import ttk
from tkinter import PhotoImage
```

### Data Preprocessing
Getting data from database

```python
def sort_genres(genres_str):    # sorting genre based on alphabetical order
    genres_list = genres_str.split(', ')
    sorted_genres = sorted(genres_list)
    return ', '.join(sorted_genres)


def get_data():
    global df
    df = pd.read_csv("new_anime.csv",encoding="utf-8") #modify this according to your file directory
    df["MAL_ID"] = df.index   # converting MAL_ID to match the index
    df["Genres"] = df["Genres"].apply(sort_genres)
```

### Editing Database
Adding or deleting entries

```python
def delete_entry():
    global df
    removed_ID = int(input("enter MAL_ID to be deleted: "))
    print(f"you are trying to remove the following entry:]\n{df.iloc[removed_ID]}")
    confirmation = input("confirm to remove it? yes or no? ").upper()
    if confirmation == "YES":
        df = df.drop(index=removed_ID)
    elif confirmation != "YES" and confirmation != "NO":
        print("You have entered an invalid entry")
    df.to_csv("new_anime.csv", index=False)


# for dev to add entries to the database
def add_entry():
    newID = df.iloc[len(df) - 1, 0] + 1
    newName = input("enter anime name: ").capitalize()
    newScore = float(input("enter rating of anime: "))
    newGenres = input("enter Genres, separated by commas: ").title()
    newGenres = sort_genres(newGenres)  # sorting of entered genres
    newSynopsis = input("enter synopsis: ")
    df.loc[len(df)] = [newID, newName, newScore, newGenres, newSynopsis]
    df.to_csv("new_anime.csv", index=False)
```

### Search functions


These functions searches based on variables such as name, score and genres

```python
def search_by_genres(search_query):
    global df
    genres_list = [genre.strip() for genre in search_query.split(',')]
    pattern = r'(?=.*' + r')(?=.*'.join(genres_list) + r')'
    df = df[df["Genres"].str.contains(pattern, flags=re.IGNORECASE, na=False)]
    return df


def search_by_min_score(search_query):
    global df
    search_query = float(search_query)
    df = df.loc[df["Score"] >= search_query]
    return df


def search_by_max_score(search_query):
    global df
    search_query = float(search_query)
    df = df.loc[df["Score"] <= search_query]
    return df


def search_by_name(search_query):
    global df
    names_list = [name.strip() for name in search_query.split(',')]
    pattern = r'(?=.*' + r')(?=.*'.join(names_list) + r')'
    df = df[df["Name"].str.contains(pattern, flags=re.IGNORECASE, na=False)]
    return df
```

### Sorting functions
Sort the database based on scores

```python
def sort_by_increasing_score():
    global df
    df = df.sort_values("Score")
    return df


# sorting based on decreasing score
def sort_by_decreasing_score():
    global df
    df = df.sort_values("Score", ascending = False)
    return df
```

# User Interface

Designing buttons used in user interface

```python
def btn_cancel_clicked():
    window.destroy()
    quit_window()


def btn_reset_clicked():
    window.destroy()
    startup()


def btn_exit_clicked():
    exit_window.destroy()


def btn_search_clicked():
    selected=drop_down.get()
    search_results.delete(1.0,tk.END) # this function clears the previous search result
    if selected == "Filter by...":
        search_query = search_entry.get()
        search_results.insert(tk.END, "please select an option!")

    if selected == "Title":
        search_query = search_entry.get()
        search_by_name(search_query)
        search_results.insert(tk.END, df.to_string())

    elif selected == "Minimum Score":
        search_query = search_entry.get()
        search_by_min_score(search_query)
        search_results.insert(tk.END, df.to_string())

    elif selected == "Maximum Score":
        search_query = search_entry.get()
        search_by_max_score(search_query)
        search_results.insert(tk.END, df.to_string())

    elif selected == "Genres":
        search_query = search_entry.get()
        search_by_genres(search_query)
        search_results.insert(tk.END, df.to_string())

    elif selected == "Lowest to Highest Score":
        sort_by_increasing_score()
        search_results.insert(tk.END, df.to_string())

    elif selected == "Highest to Lowest Score":
        sort_by_decreasing_score()
        search_results.insert(tk.END, df.to_string())

    elif selected == "First Ten Results":
        search_results.insert(tk.END, df.head(10).to_string())

    elif selected == "Random Rec(repeatable)":
        search_results.insert(tk.END, df.sample().to_string())

    else:
        search_results.insert(tk.END, "An Error has occurred, please reselect an option")
```

The main menu, allow user to interact and search based on criteria

```python
def startup():
    global window       # create the main window title
    window = tk.Tk()
    window.title("Welcome to anime.com!")
    window.geometry("1600x800")
    window.grid_rowconfigure(0, weight=1, minsize=20)
    window.grid_rowconfigure(2, weight=1, minsize=20)
    window.grid_columnconfigure(0, weight=1, minsize=20)
    window.grid_columnconfigure(2, weight=1, minsize=20)

    background_image = PhotoImage(file="anime-png-7598.png")
    background_making = tk.Label(window, image=background_image)
    background_making.place(x=0, y=0, relwidth=2, relheight=1)

    global drop_down        # for user to select from list
    options = [
        "Filter by...",
        "Title",
        "Minimum Score",
        "Maximum Score",
        "Genres",
        "Lowest to Highest Score",
        "Highest to Lowest Score",
        "First Ten Results",
        "Random Rec(repeatable)"
    ]
    drop_down = ttk.Combobox(window, width=30, values=options)  # dropdown box
    drop_down.current(0)
    drop_down.grid(row=0, column=1, sticky=tk.N, padx=5, pady=5)
    drop_down.config(background="#20252b")

    lbl = tk.Label(window, text="How would you like to filter results by? ")        # display msg for search
    lbl.grid(row=0, column=0, sticky=tk.W, padx=5, pady=5)

    global search_entry     # add spinbox for searchbar
    search_entry = tk.Entry(window, width=80)
    search_entry.grid(row=0, column=1, sticky=tk.W, padx=5, pady=5)

    # adding buttons
    btnSearch = tk.Button(window, text="Search", command=btn_search_clicked)  # search button
    btnSearch.grid(row=1, column=0)

    btnCancel = tk.Button(window, text="Exit Menu", command=btn_cancel_clicked)  # cancel button
    btnCancel.grid(row=1, column=1, sticky=tk.N, padx=5, pady=5)

    btnReset = tk.Button(window, text="Reset all filters", command=btn_reset_clicked)  # reset button
    btnReset.grid(row=1, column=2, sticky=tk.N, padx=5, pady=5)

    # search results
    global search_results
    search_results = tk.Text(window, wrap=tk.WORD, height=35, width=150)  # results of search input in a box
    search_results.grid(row=2, column=1)

    window.mainloop()
```

Exit menu

```python
def quit_window():
    global exit_window
    exit_window = tk.Tk()
    pikachu = PhotoImage(file="pikachupngfile.png")
    thank_you_msg = tk.Label(exit_window,
                             text="Thank You For Using Our Program",
                             font=(70),
                             relief="raised",
                             image=pikachu,
                             compound="bottom"
                             )
    thank_you_msg.pack()
    btnExit = tk.Button(exit_window, text = "Exit", command = btn_exit_clicked, compound="bottom")
    btnExit.pack()
    exit_window.mainloop()
```

Executing of window
